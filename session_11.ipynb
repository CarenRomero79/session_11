## 1. Valor: maximizar el valor

---



El objetivo financiero principal de un proyecto es maximizar el valor que genera la inversión. Esto significa buscar que los recursos invertidos produzcan beneficios económicos superiores a los costos y al capital utilizado.

Maximizar el valor implica analizar:

Ingresos esperados: cuánto dinero generará el proyecto.
Costos: cuánto cuesta desarrollar y operar el proyecto.
Inversión inicial: recursos necesarios para ponerlo en funcionamiento.
Flujos de caja: dinero que realmente entra y sale durante el proyecto.
Riesgo: posibilidad de que los resultados sean diferentes a los esperados.
Tiempo: cuándo se recupera la inversión y cuándo comienzan a generarse beneficios.

Por ejemplo, si una empresa invierte $100 millones en un proyecto y espera obtener flujos de caja que permitan recuperar la inversión y generar $30 millones adicionales de valor, el proyecto estaría generando valor económico.

Idea clave: no basta con que un proyecto genere ingresos; debe generar valor después de cubrir sus costos, inversión y riesgo.



## 2. CAPEX vs. OPEX

---



Una de las primeras clasificaciones financieras de un proyecto es diferenciar entre CAPEX y OPEX.

CAPEX – Gastos de capital

CAPEX significa Capital Expenditures o gastos de capital.

Son los recursos destinados a adquirir, construir o mejorar activos que tendrán una utilidad durante varios períodos.

Ejemplos:

Compra de maquinaria.
Equipos tecnológicos.
Servidores.
Construcción o adecuación de instalaciones.
Compra de vehículos.
Implementación inicial de infraestructura.
Sistemas tecnológicos de larga duración.

Normalmente se concentran principalmente en la etapa inicial del proyecto.

OPEX – Gastos operativos

OPEX significa Operating Expenses o gastos operativos.

Son los gastos necesarios para mantener funcionando el proyecto o negocio.

Ejemplos:

Salarios.
Servicios públicos.
Mantenimiento.
Internet.
Licencias periódicas.
Arrendamiento.
Insumos.
Soporte técnico.
Publicidad y administración.
Diferencia principal
CAPEX	OPEX
Inversión en activos	Gastos para operar
Generalmente ocurre al inicio	Ocurre durante la operación
Genera beneficios durante varios períodos	Permite mantener la operación
Ejemplo: comprar un servidor	Ejemplo: pagar mantenimiento
Se relaciona con inversión	Se relaciona con operación

Ejemplo: si una empresa implementa un sistema de análisis de datos, comprar los servidores puede considerarse CAPEX, mientras que pagar mensualmente el almacenamiento en la nube y soporte puede considerarse OPEX.

## 3. Métricas de viabilidad financiera

---



Las métricas financieras permiten determinar si un proyecto es económicamente viable y si la inversión genera suficientes beneficios.

VAN – Valor Actual Neto

El VAN (Valor Actual Neto) permite determinar cuánto valor genera actualmente un proyecto teniendo en cuenta el valor del dinero en el tiempo.

La idea es traer los flujos de caja futuros al valor presente y compararlos con la inversión inicial.

Una representación simplificada es:

VAN = Valor presente de los flujos futuros − Inversión inicial

Interpretación general:

VAN > 0: el proyecto genera valor por encima de la tasa de descuento utilizada.
VAN = 0: el proyecto genera exactamente el rendimiento exigido.
VAN < 0: el proyecto no alcanza el rendimiento exigido bajo los supuestos utilizados.

El VAN es especialmente importante porque $1 hoy no tiene el mismo valor que $1 recibido dentro de varios meses o años.

TIR – Tasa Interna de Retorno

La TIR es la tasa de rentabilidad que hace que el VAN sea igual a cero.

En términos sencillos, representa el rendimiento implícito de los flujos de caja del proyecto.

La TIR se puede comparar con una tasa de referencia o tasa mínima requerida.

Por ejemplo:

TIR = 18 %
Tasa mínima requerida = 12 %

La diferencia indica que, bajo esos supuestos, el proyecto genera una rentabilidad superior a la tasa mínima requerida.

Importante: la TIR debe analizarse junto con el VAN y los flujos de caja; por sí sola no siempre es suficiente para evaluar un proyecto.

ROI – Retorno sobre la inversión

El ROI (Return on Investment) mide cuánto beneficio se obtiene en relación con la inversión realizada.

Una fórmula básica es:

ROI = (Beneficio obtenido − Inversión) / Inversión × 100

Ejemplo:

Si se invierten $100 millones y se obtiene un beneficio de $130 millones:

ROI = (130 − 100) / 100 × 100 = 30 %

Esto significa que el beneficio obtenido representa un retorno del 30 % sobre la inversión inicial, según esta definición.

Recuperación de la inversión

También se puede analizar el período de recuperación (Payback).

Este indicador responde a:

¿Cuánto tiempo necesitamos para recuperar el dinero invertido?

Por ejemplo:

Inversión inicial: $100 millones

Flujo neto mensual:

Mes 1: $20 millones
Mes 2: $25 millones
Mes 3: $30 millones
Mes 4: $25 millones

Acumulado:

Mes 1: $20 millones
Mes 2: $45 millones
Mes 3: $75 millones
Mes 4: $100 millones

La inversión se recuperaría aproximadamente en 4 meses.

## 4. Elementos del modelo de flujo de caja

---



Para evaluar financieramente el proyecto necesitamos construir un modelo de flujo de caja.

El modelo muestra cuánto dinero entra y cuánto dinero sale durante cada período.

Ingresos

Son los recursos económicos que recibe el proyecto.

Pueden provenir de:

Ventas.
Servicios.
Nuevos clientes.
Ahorros generados por el proyecto.
Reducción de costos.
Nuevas fuentes de ingresos.

Por ejemplo, si un proyecto permite aumentar las ventas en $20 millones mensuales, ese incremento puede formar parte de los ingresos o beneficios económicos del proyecto.

Costos

Son los recursos necesarios para desarrollar y operar el proyecto.

Pueden dividirse en:

Costos de inversión

Relacionados principalmente con CAPEX.

Ejemplo:

Equipos: $50 millones.
Implementación: $20 millones.
Infraestructura: $30 millones.

Inversión inicial = $100 millones.

Costos operativos

Relacionados principalmente con OPEX.

Ejemplo:

Personal: $5 millones/mes.
Software: $2 millones/mes.
Mantenimiento: $1 millón/mes.

OPEX mensual = $8 millones.

Flujo neto

El flujo neto representa la diferencia entre los recursos que entran y los que salen.

Una fórmula sencilla es:

Flujo neto = Ingresos − Costos − Inversiones

Ejemplo:

Ingresos = $30 millones
Costos = $10 millones
Inversión = $5 millones

Flujo neto = $30 − $10 − $5 = $15 millones

El flujo neto es fundamental para calcular el VAN, la TIR y el período de recuperación.

Tasa de descuento

La tasa de descuento permite convertir los flujos futuros a su equivalente actual.

Esto es necesario porque recibir $10 millones hoy no es financieramente igual que recibir $10 millones dentro de un año.

La tasa puede incorporar factores como:

Costo del capital.
Riesgo del proyecto.
Inflación, según el modelo utilizado.
Alternativas de inversión.
Condiciones financieras de la empresa.

Por ejemplo, si se utiliza una tasa de descuento del 12 % anual, los flujos futuros se descuentan utilizando esa tasa para calcular su valor presente.

Modelo de flujo de caja a 14 meses

Si el proyecto tiene un horizonte de 14 meses, se puede construir una tabla de este tipo:

Período	Ingresos	CAPEX	OPEX	Flujo neto	Flujo acumulado
Mes 0	$0	$100 M	$0	-$100 M	-$100 M
Mes 1	$20 M	$0	$8 M	$12 M	-$88 M
Mes 2	$25 M	$0	$8 M	$17 M	-$71 M
Mes 3	$30 M	$0	$8 M	$22 M	-$49 M
...	...	...	...	...	...
Mes 14	$X	$0	$X	$X	$X

El mes 0 representa normalmente la inversión inicial.

A partir del mes 1 se proyectan los ingresos y costos.

Con esta información se pueden calcular:

Flujo neto mensual.
Flujo acumulado.
Período de recuperación.
VAN.
TIR.
ROI.
Rentabilidad del proyecto.
¿Por qué 14 meses?

En este caso, los 14 meses representan el horizonte de evaluación. Se puede analizar si durante ese período el proyecto recupera la inversión y empieza a generar valor.

Es importante diferenciar:

Horizonte del proyecto = 14 meses

de

Período de recuperación = número de meses necesarios para recuperar la inversión.

Por ejemplo, un proyecto puede evaluarse durante 14 meses pero recuperar su inversión en el mes 9.

**## 5. Entregables financieros clave**

Al finalizar el análisis financiero, se deben presentar documentos o resultados que permitan tomar una decisión informada.

CAPEX.
Implementación.
Equipos.
Tecnología.
Otros gastos iniciales.
2. Presupuesto operativo

Muestra los costos necesarios para mantener el proyecto funcionando.

Incluye:

OPEX.
Personal.
Mantenimiento.
Servicios.
Licencias.
Administración.
3. Modelo de flujo de caja

Es uno de los principales entregables.

Debe contener:

Ingresos → Costos → Flujo neto → Flujo acumulado

Debe mostrar:

Tasa de descuento utilizada.
Flujos proyectados.

Interpretación del resultado.

Debe presentar:

Flujos de caja.
TIR calculada.
Tasa mínima requerida.
Comparación entre ambas.
6. ROI

Debe mostrar:

Inversión realizada.
Beneficio obtenido.
ROI porcentual.
